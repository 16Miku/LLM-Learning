For longer text generations, you can stream the response. This displays the output token by token as it's generated, rather than waiting for the entire response to complete. This provides a more interactive and responsive experience. To enable this, simply set stream=True.

对于更长的文本生成，您可以流式传输响应。这将逐个显示输出令牌，而不是等待整个响应完成。这提供了更交互和响应更快的体验。要启用此功能，只需将 stream 设置为 True。

```python
# @title Simple streaming example
from google.colab import ai

stream = ai.generate_text("介绍斋藤飞鸟", stream=True)
for text in stream:
  print(text, end='')
```

```python
#@title Text formatting setup
# 此代码不是 `colab.ai` 库必需的，但对于格式化文本块非常有用。
# 这个单元格的主要作用是定义一个工具类 `LineWrapper`，用于后续对流式文本进行格式化。
# 因此，单独执行这个单元格并不会产生可见的输出，它的“输出”是成功定义了 `LineWrapper` 这个类。
# 上一个文本单元格 `ysDdFbH_Dgtz` 介绍了“流式传输”的概念，即逐个令牌显示输出，提供更交互的体验。
# 这个 `LineWrapper` 类就是为了优化这种流式传输的显示效果，使其在控制台中更易读，避免文本溢出单行。
# 随后在 `DWiLPzTnRoy-` 单元格中，它被实例化并用于格式化 `ai.generate_text` 的流式输出。

import sys

# 定义一个名为 LineWrapper 的类，用于将文本格式化为指定行宽。
class LineWrapper:
    # 构造函数，在创建 LineWrapper 对象时调用。
    def __init__(self, max_length=80):
        # 初始化最大行长，默认为80个字符。这意味着每行最多显示80个字符。
        self.max_length = max_length
        # 初始化当前行已有的字符长度，用于跟踪当前行的内容长度。
        # 每当打印字符时，这个值会更新，用于判断是否需要换行。
        self.current_line_length = 0

    # 定义一个名为 print 的方法，用于逐块打印文本并进行换行。
    # 这个方法接收一个 `text_chunk`（文本片段），并尝试将其格式化打印。
    def print(self, text_chunk):
        # 初始化索引i，用于遍历文本块。从文本块的开始位置开始处理。
        i = 0
        # 获取文本块的长度，以便知道何时停止遍历。
        n = len(text_chunk)
        # 循环直到遍历完整个文本块。
        while i < n:
            # 记录当前单词的起始索引。每次处理一个新单词时，都会记录它的开始位置。
            start_index = i
            # 查找当前单词的末尾，直到遇到空格或换行符。
            # 这个循环的目的是识别一个完整的“单词”或一个标点符号序列，不包括分隔符。
            while i < n and text_chunk[i] not in ' \n': # 寻找单词的末尾
                i += 1
            # 提取当前单词。`current_word` 变量现在存储了从 `start_index` 到 `i-1` 的字符。
            current_word = text_chunk[start_index:i]

            # 初始化分隔符为空字符串。这是为了后续捕获单词后面的空格或换行符。
            delimiter = ""
            # 如果i没有超出文本块的范围，说明找到了一个分隔符。
            if i < n: # 如果不是块的末尾，我们找到了一个分隔符
                # 获取分隔符（空格或换行符）。
                delimiter = text_chunk[i]
                i += 1 # 消费分隔符，将索引i向前移动一位，准备处理下一个单词。

            # 如果当前单词不为空，则进行进一步处理。
            if current_word:
                # 判断当前行在写入新单词前是否需要一个前导空格。
                # 如果当前行已经有内容（长度大于0），则通常需要一个空格来分隔单词。
                needs_leading_space = (self.current_line_length > 0)

                # 情况1: 单词本身就比最大行长还要长（必须被拆分）。
                # 例如，一个非常长的URL或没有空格的长字符串。
                if len(current_word) > self.max_length:
                    # 如果当前行有内容，先换行，确保长单词从新行开始。
                    if needs_leading_space: # 如果当前行有内容，则换行
                        sys.stdout.write('\n')
                        self.current_line_length = 0
                    # 逐字符打印这个长单词，并在达到最大行长时强制换行。
                    for char_val in current_word: # 拆分长单词
                        if self.current_line_length >= self.max_length:
                            sys.stdout.write('\n')
                            self.current_line_length = 0
                        sys.stdout.write(char_val) # 打印单个字符
                        self.current_line_length += 1 # 更新当前行长
                # 情况2: 单词不适合放在当前行（即使加上一个空格也会超过最大行长），因此在新行打印。
                elif self.current_line_length + (1 if needs_leading_space else 0) + len(current_word) > self.max_length:
                    sys.stdout.write('\n') # 换行
                    sys.stdout.write(current_word) # 打印单词
                    self.current_line_length = len(current_word) # 更新当前行长为新单词的长度
                # 情况3: 单词适合放在当前行。
                else:
                    # 如果需要前导空格。
                    if needs_leading_space:
                        # 定义一组不应该有前导空格的标点符号。
                        # 当这些标点符号（如逗号、句号）紧跟在一个单词后面时，不需要额外的空格。
                        no_leading_space_punctuation = {
                            ",", ".", ";", ":", "!", "?",        # 标准句子标点
                            ")", "]", "}",                     # 闭合括号
                            "'s", "'S", "'re", "'RE", "'ve", "'VE", # 常见缩写
                            "'m", "'M", "'ll", "'LL", "'d", "'D",
                            "n't", "N'T",
                            "...", "…"                          # 省略号
                        }
                        # 如果当前单词不是上述标点符号，则添加一个空格。
                        # 这样可以避免在标点符号前出现不必要的空格（如“word ,”）。
                        if current_word not in no_leading_space_punctuation:
                            sys.stdout.write(' ')
                            self.current_line_length += 1
                    sys.stdout.write(current_word) # 打印单词
                    self.current_line_length += len(current_word) # 更新当前行长

            # 如果分隔符是换行符。
            if delimiter == '\n':
                sys.stdout.write('\n') # 强制换行，开始新的一行。
                self.current_line_length = 0 # 重置当前行长，因为已经换行。
            # 如果分隔符是空格。
            elif delimiter == ' ':
                # 如果当前行已满并且遇到了空格分隔符，意味着需要换行。
                # 即使空格本身不占用显示空间，但它表示单词间的分隔，若当前行已到最大长度，应换行。
                if self.current_line_length >= self.max_length:
                    sys.stdout.write('\n')
                    self.current_line_length = 0

        sys.stdout.flush() # 刷新输出缓冲区，确保所有内容都被立即写入标准输出。这对于流式输出非常重要，可以确保用户能实时看到文本生成的过程。
```

```python
# @title Formatted streaming example
from google.colab import ai

wrapper = LineWrapper()
for chunk in ai.generate_text('生成一份详细的关于斋藤飞鸟的报告', model_name='google/gemini-2.5-flash', stream=True):
  wrapper.print(chunk)
```
